```python
# RRCP (19/09/2026): Derivadas
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Funcion
def f(x):
    return x**2
```

```python
# Deribada de la funcion
def df(f):
    h = 0.000001
    return (f(x + h) - f(x))/h 
```

```python
# Graficar la funcion.
x = np.linspace(-10, 10, num = 1000)
y = f(x)
plt.plot(x,y)
plt.grid()
```

```python
# Graficar la funcion con la derivada
plt.plot(x, y, 'b') # Azul
plt.plot(x, df(f), 'r') # Rojo
plt.grid()
```
